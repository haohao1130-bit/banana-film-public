**全新前端・獨立製片流程**：不再在 Colab 裡一格一格編輯劇情與角色。完成模型安裝後，開啟網頁即可：填完整提示詞 → 上傳參考圖 → 全片 9B 備稿 → 展開檢查所有 H3 拍攝稿 → 確認 H3 開拍 → 播放／續拍。

程式公開，不需 GitHub Token。請勿與其他 MV／短片任務在同一個 ComfyUI 同時生成。


```python
#@title ① 安裝 VNext 全新前端與 ComfyUI
from pathlib import Path
import os, subprocess, sys
PROJECT = Path('/content/banana-film-vnext-web')
BRANCH = 'feature/public-film-vnext-isolated-20261109'
REPO = 'https://github.com/haohao1130-bit/banana-film-public.git'
if not PROJECT.exists():
    subprocess.run(['git','-c','credential.helper=','clone','--depth','1','--branch',BRANCH,REPO,str(PROJECT)],check=True)
elif not (PROJECT/'tools/vnext_runtime.py').is_file():
    raise RuntimeError('舊程式目錄不相容，請重啟 Colab 執行階段。')
sys.path.insert(0,str(PROJECT))
sys.path.insert(0,str(PROJECT/'tools'))
import vnext_runtime as vnext
vnext.install(globals())
print('✅ VNext 程式和 ComfyUI 安裝完成')
```

```python
#@title ② 下載 MiniMax H3 + 本地 9B 導演模型
USE_DRIVE_CACHE = False #@param {type:"boolean"}
vnext.download(globals(), use_drive_cache=USE_DRIVE_CACHE)
print('✅ H3 與 9B 模型備妥')
```

```python
#@title ③ 啟動 ComfyUI，開啟 VNext 完整網頁前端
SAVE_TO_DRIVE = True #@param {type:"boolean"}
vnext.launch(globals(), save_to_drive=SAVE_TO_DRIVE)
```
